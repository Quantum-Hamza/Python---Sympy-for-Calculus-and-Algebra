## **¡Aguanta, amigo!**
Exploremos ahora el cálculo infinitesimal con ***sympy***: **derivadas** e **integrales**.

```python
# Derivada de la función dada respecto de la variable dada
f = exp(x**2)
diff(f, x)
```

```python
# Podemos pedir derivadas de orden superior (el tercer argumento denota el orden)
diff(x**4, x, 3)
```

```python
# Al igual que con la integral (vista anteriormente), Derivative() muestra la derivada pero no la calcula
deriv = Derivative(x**4, x, 3)
deriv
```

```python
# Podemos calcularla con doit()
deriv.doit()
```

```python
# Podemos calcular integrales indefinidas (sin la constante)
integrate(cos(x), x)
```

```python
integrate(tan(x), x)
```

```python
# Podemos calcular integrales impropias
f = exp(-x)
integrate(f, (x, 0, oo))
```

```python
# Como vimos antes con Integral()...
expinf = Integral(f, (x, 0, oo))
expinf
```

```python
# Y aplicando doit()
expinf.doit()
```

```python
# Si sympy no la puede calcular, simplemente la declara
integrate(x**x, x)
```

## Ahora, ¡resolvamos ecuaciones!

```python
x, y, z = symbols('x, y, z')
# Eq sirve para igualar los dos miembros
Eq(x, y)
```

```python
# solveset() nos permite resolver ecuaciones
solveset(Eq(x**2, 1), x)
```

```python
# Y de este modo
solveset(x**2 - 1, x)
```

```python
# Por defecto, la ecuación se resuelve en los complejos
solveset(x**2 + 1, x)
```

```python
# Podemos calcular solo dentro de los reales:
solveset(x**2 + 1, x, domain=S.Reals)
```

```python
solveset(sin(x)**2 + cos(x)**2 - 1, x, domain=S.Reals)
```

```python
solveset(sin(x) - 1, x, domain=S.Reals)
```

```python
solveset(exp(x), x)
```

```python
# Esta ecuación no tiene solución analítica, por lo que sympy simplemente reescribe lo mismo
solveset(cos(x) - x, x)
```

## ¡Ah! ¡Que también puedo resolver sistemas!

```python
# Para hacerlo, usaremos linsolve()
ecs = [x + y + z - 1, x + y + 2*z - 3]
linsolve(ecs, (x, y, z))
```

```python
# Podemos resolverlo con la matriz ampliada
linsolve(Matrix([[1, 1, 1, 1], [1, 1, 2, 3]]), (x, y, z))
```

```python
# Podemos dar el sistema en forma A*x = b
A, b = linear_eq_to_matrix(ecs, (x, y, z))
system = A, b
linsolve(system, (x, y, z))
```
